```python
from langgraph.store.memory import InMemoryStore
```

```python
from langchain_community.embeddings.fastembed import FastEmbedEmbeddings
embedding_model = FastEmbedEmbeddings()
```

```python
store = InMemoryStore(
    index = {
        "embed": embedding_model,
        "dims":768
    }
)
```

```python
namespace = ("user","u1")
namespace2 = ("user","u2")
```

```python
store.put(
    namespace=namespace,
    key = "name",
    value="my name is nirzar"
    )
store.put(
    namespace=namespace,
    key = "age",
    value="my age is 21"
    )
store.put(
    namespace=namespace,
    key = "gender",
    value="i am male"
    )
```

```python
store.put(
    namespace=namespace,
    key="bio",
    value="Hi, I am Khushi. I am passionate about technology, design, and creating meaningful digital experiences. I enjoy learning new skills, exploring creative ideas, and building projects that solve real-world problems."
)

store.put(
    namespace=namespace,
    key="education",
    value="I am currently pursuing my Bachelor's degree in Computer Science. My coursework includes Data Structures, Machine Learning, Databases, and Web Development."
)

store.put(
    namespace=namespace,
    key="skills",
    value="I have skills in Python, JavaScript, React, FastAPI, and basic DevOps. I also enjoy working with UI/UX design and improving user experiences."
)

store.put(
    namespace=namespace,
    key="hobbies",
    value="In my free time, I like reading books, listening to music, traveling to new places, and exploring cafes. I also enjoy sketching and journaling."
)

store.put(
    namespace=namespace,
    key="goals",
    value="My goal is to become a skilled software engineer and work on impactful products. I also want to contribute to open-source and continuously improve my technical and problem-solving skills."
)

store.put(
    namespace=namespace,
    key="personality",
    value="I am a curious, creative, and optimistic person. I enjoy collaborating with others and learning from different perspectives."
)

store.put(
    namespace=namespace,
    key="experience",
    value="I have worked on multiple academic and personal projects including web applications, APIs, and small AI-based tools. I am continuously improving my coding and system design skills."
)

store.put(
    namespace=namespace,
    key="favorite_technologies",
    value="I enjoy working with modern technologies like React for frontend, FastAPI for backend, and Docker for containerization."
)

store.put(
    namespace=namespace,
    key="daily_routine",
    value="My daily routine includes studying, coding, attending classes, and spending some time learning new technologies or working on personal projects."
)

store.put(
    namespace=namespace,
    key="contact_preferences",
    value="I prefer communication via email or professional platforms like LinkedIn. I value clear and respectful communication."
)
```

```python
store.put(
    namespace=namespace2,
    key="name",
    value="my name is Khushi"
)

store.put(
    namespace=namespace2,
    key="age",
    value="my age is 21"
)

store.put(
    namespace=namespace2,
    key="gender",
    value="i am female"
)
```

```python
store.get(
    namespace=namespace,
    key = "name"
)
```

```python
items = store.search(namespace)

for item in items:
    print(item.key,":",item.value)
```

```python
items = store.embeddings    

for item in items:
    print(item)
```

```python
items = store.search(namespace,query="what is my name",limit=4)

for item in items:
    print(item.value)
```
